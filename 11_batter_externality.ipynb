###  The batter externality, swept over `phi`
`phi` is the share of the inventory cost a batter perceives. I set it to 1/9 earlier as a round-number benchmark, not an estimate, so here it is varied. Each run is the next draw from the same seeded stream, as in the paper, so small differences between rows include simulation noise as well as the effect of `phi`.

```python
from abs_common import *           
```

```python
run = game_model_run()
PHIS = [1.0, 0.5, 0.25, 1/9, 0.0]
rows, SWEEP = [], {}
for phi in PHIS:
    s = run[phi]; SWEEP[phi] = s
    b, d = s["ok"][s["is_b"]].mean(), s["ok"][~s["is_b"]].mean()
    rows.append(dict(phi=round(phi, 3), batter_rate=pct(b), defense_rate=pct(d),
                     gap_def_minus_bat_pts=round(100*(d-b), 1),
                     share_of_observed_gap=pct((d-b)/GAP_OBS),
                     team_overturns_per_game=round(s["ok"].sum()/s["N"], 3)))
sweep = pd.DataFrame(rows)
print(sweep.to_string(index=False))
base_wins = SWEEP[1.0]["ok"].sum() / SWEEP[1.0]["N"]
phi9 = SWEEP[1/9]["ok"].sum() / SWEEP[1/9]["N"]
cost_pct = 100 * (1 - phi9 / base_wins)
print(f"\nObserved defense-batter gap: {100*GAP_OBS:.1f} points")
print(f"Cost of the externality at phi=1/9: {cost_pct:.1f}% of a team's expected overturns, "
      f"about {(base_wins-phi9)*162:.1f} per 162 games, or {(base_wins-phi9)*162*RUNS[1]:.2f} runs at {RUNS[1]} runs each")
```
