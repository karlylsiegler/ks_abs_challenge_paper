### same count, different challenger
Defense and batter rates are not directly comparable if they contest different pitches. Comparing them within the same ball-strike count removes one source of difference, though not all of it (the two sides still mostly contest called balls versus called strikes).

```python
from abs_common import *         
```

```python
cnt = COUNT.copy()
cnt["gap_def_minus_bat"] = (cnt.pct_def - cnt.pct_bat).round(1)
print(cnt[["balls","strikes","n_bat","pct_bat","n_def","pct_def","gap_def_minus_bat"]].to_string(index=False))

# Batter success by strikes, weighted by challenge counts
cnt["ok_bat"] = cnt.n_bat * cnt.pct_bat / 100
cnt["ok_def"] = cnt.n_def * cnt.pct_def / 100
st = cnt.groupby("strikes")[["n_bat","ok_bat","n_def","ok_def"]].sum()
st["batter_rate"] = (100*st.ok_bat/st.n_bat).round(1)
st["defense_rate"] = (100*st.ok_def/st.n_def).round(1)
st["share_of_batter_chal"] = (100*st.n_bat/st.n_bat.sum()).round(1)
print()
print(st[["n_bat","batter_rate","defense_rate","share_of_batter_chal"]].to_string())
```
