```python
def row(label, n, k):
    lo, hi = wilson(k, n)
    return dict(role=label, challenges=int(n), overturned=int(k), rate=pct(k/n), ci_95=f"{pct(lo)}-{pct(hi)}")

Cn, Ck = ROLE.loc["catcher"]; Bn, Bk = ROLE.loc["batter"]; Pn, Pk = ROLE.loc["pitcher"]
Dn, Dk = Cn + Pn, Ck + Pk
roles = pd.DataFrame([row("catcher",Cn,Ck), row("pitcher",Pn,Pk), row("defense (C+P)",Dn,Dk),
                      row("batter",Bn,Bk), row("all",TOTAL,ROLE.ot.sum())])
print(roles.to_string(index=False))
GAP_OBS = Dk/Dn - Bk/Bn
Z_P3 = ztest(Dk, Dn, Bk, Bn)
print(f"\nDefense minus batter: {100*GAP_OBS:.1f} points, z = {Z_P3:.1f}")
print(f"Catcher minus batter: {100*(Ck/Cn-Bk/Bn):.1f} points, z = {ztest(Ck,Cn,Bk,Bn):.1f}")
print(f"Catchers make {pct(Cn/Dn)} of defensive challenges and {pct(Cn/TOTAL)} of all challenges")
```
