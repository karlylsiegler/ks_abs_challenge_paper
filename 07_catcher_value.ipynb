### how much is the catcher's challenge skill worth?
**Method.** Apply the batter success rate to the catchers' challenges and compare with their actual successes. This is a rough benchmark, not a controlled comparison: batters and catchers contest different pitches, so the batter rate is an imperfect stand-in.

**ASSUMED:** the run value of an overturned call. A flat value is a simplification, because an overturn is worth very different amounts by base-out state, count and game context. Baseball Savant has situational run values that I did not have at pitch level, so the table is a sensitivity range, not an estimate.

```python
from abs_common import *           
```

```python
extra = Ck - Cn * Bk / Bn               # catcher overturns beyond a batter-level success rate
RUNS = [0.08, 0.12, 0.16]               # ASSUMED runs per overturned call
TEAMS, RUNS_PER_WIN = 30, 10            # ASSUMED conversion
value = pd.DataFrame({"runs_per_overturn": RUNS,
                      "league_runs": [extra*r for r in RUNS],
                      "per_team_runs": [extra*r/TEAMS for r in RUNS],
                      "per_team_wins": [extra*r/TEAMS/RUNS_PER_WIN for r in RUNS]}).round(2)
print(f"Catchers made {pct(Cn/TOTAL)} of challenges and produced {pct(Ck/ROLE.ot.sum())} of overturns.")
print(f"Extra overturns vs. a batter-level rate: {extra:.0f} (about {extra/TEAMS:.0f} per team)\n")
print(value.to_string(index=False))

# Individual evaluation: Millas won 22 of 51 (ESPN), the lowest rate among catchers with 50+
mk, mn = MILLAS
exp_wins = mn * Ck / Cn
mlo, mhi = wilson(mk, mn)
print(f"\nMillas: {mk}/{mn} = {pct(mk/mn)}; league catcher rate implies {exp_wins:.1f} wins, a shortfall of {exp_wins-mk:.1f}")
print(f"95% interval {pct(mlo)}-{pct(mhi)} vs league catcher rate {pct(Ck/Cn)}")
print(f"Cost of that shortfall at {RUNS[1]} runs per overturn: {(exp_wins-mk)*RUNS[1]:.1f} runs")
```
